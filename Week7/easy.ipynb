```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# EASY: Deutsch Algorithm for constant function f(x) = 0
# ---------------------------------------------------------

# Create simulator
simulator = AerSimulator()

# Create circuit
# q0 = input qubit
# q1 = output/ancilla qubit
qc = QuantumCircuit(2, 1)

# Step 1: Prepare input qubit in |+>
qc.h(0)

# Step 2: Prepare ancilla in |->
qc.x(1)
qc.h(1)

# Step 3: Oracle for f(x) = 0
# Since f(x) is always 0, oracle does nothing.

# Step 4: Apply Hadamard to input qubit
qc.h(0)

# Step 5: Measure input qubit
qc.measure(0, 0)

# Display circuit
print("Deutsch Circuit - Constant Function f(x)=0")
display(qc.draw("mpl"))

# Run circuit
result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()

print("Measurement results:", counts)

# Plot histogram
plot_histogram(counts)
plt.show()

# Determine result
if counts.get("0", 0) > counts.get("1", 0):
    print("Result: 0 → CONSTANT")
else:
    print("Result: 1 → BALANCED")
```
