```python
# Importing the necessary Qiskit modules
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_state_city
from IPython.display import display

# 1. Initialise a Quantum Circuit with 2 qubits
# No classical bits are needed since we are inspecting the mathematical state directly
qc = QuantumCircuit(2)

# 2. Build the arbitrary entangled state
# An Hadamard gate rotates by pi/2. Here, we rotate by 2*pi/3 to create an unequal split.
# This puts qubit 0 into the state: 1/2 |0> + sqrt(3)/2 |1>
theta = 2 * np.pi / 3
qc.ry(theta, 0)

# Apply a CNOT gate to entangle the second qubit with the first
qc.cx(0, 1)

# 3. Draw the colourful circuit diagram
# The 'iqp' style gives the diagram a vibrant look in Jupyter Notebooks
print("Drawing the colourful circuit diagram...")
display(qc.draw(output='mpl', style='iqp'))

# 4. Verify the state via Statevector inspection
# Statevector simulates the exact mathematical state of the circuit without measurement collapse
state = Statevector.from_instruction(qc)

print("\nMathematical Statevector (Dirac Notation):")
display(state.draw('latex'))

print("\nProbabilities of each outcome:")
probs = state.probabilities_dict()
for outcome, prob in probs.items():
    print(f"State |{outcome}> : {prob:.2%} chance")

# 5. Visualise the state density matrix
# This acts as theoretical state tomography, mapping the real and imaginary parts
print("\nPlotting the State City (Density Matrix)...")
display(plot_state_city(state, color=['darkorange', 'midnightblue'], title="Theoretical State Tomography"))
```
