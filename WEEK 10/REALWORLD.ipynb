```python
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, pauli_error, depolarizing_error
# Classical bandwidth vs superdense-coded transmission for 100 two-bit messages
n_messages = 100
bits_per_message = 2
classical_bits = n_messages * bits_per_message # one classical bit per classical bit sent
sd_qubits_live = n_messages * 1 # one qubit sent per message, live
sd_pairs_predistributed = n_messages # one Bell pair shared in advance per message
sd_qubits_predistributed = sd_pairs_predistributed * 2
print(f"{'Quantity':<48}{'Classical':<14}{'Superdense coding'}")
print(f"{'Bits of information delivered':<48}{classical_bits:<14}{n_messages*bits_per_message}")
print(f"{'Carriers sent during the live session':<48}{str(classical_bits)+'bits':<14}{str(sd_qubits_live)+' qubits'}")
print(f"{'Information per live carrier':<48}{'1 bit/bit':<14}{'2 bits/qubit'}")
print(f"{'Extra resource needed in advance':<48}{'none':<14}{str(sd_qubits_predistributed)+'qubits (100 Bell pairs)'}")
print(f"\nLive-channel saving from superdense coding: "
    f"{classical_bits} bits -> {sd_qubits_live} qubit-sends "
    f"({classical_bits/sd_qubits_live:.0f}x fewer carriers needed at send time)")
```
